```python
sales_data = [
    (101, 1, 1000),
    (102, 2, 2000),
    (103, 1, 1500),
    (104, 3, 3000),
    (105, 2, 1000),
    (106, 5, 500)
]
```

```python
sales_df = spark.createDataFrame(
    sales_data,
    ["sale_id", "category_id", "amount"]
)
```

```python
categories_data = [
    (1, "Electronics"),
    (2, "Clothing"),
    (3, "Furniture"),
    (4, "Grocery")
]
```

```python
categories_df = spark.createDataFrame(
    categories_data,
    ["category_id", "category_name"]
)
```

```python
sales_df.show()
```

```python
categories_df.show()
```

```python
sales_summary_df = (
    sales_df
    .groupBy("category_id")
    .agg(
        sum("amount").alias("total_sales"),
        avg("amount").alias("average_sales"),
        count("*").alias("sale_count")
    ).orderBy(
        col("total_sales").desc())
)
```

```python
sales_summary_df.show()
```

```python
inner_df = sales_summary_df.join(
    categories_df, "category_id", "inner"
)

left_df = sales_df.join(
    categories_df, "category_id", "left"
)

right_df = sales_df.join(
    categories_df, "category_id", "right"
)

full_df = sales_df.join(
    categories_df, "category_id", "full"
)

print("Sales:", sales_df.count())
print("Categories:", categories_df.count())
print("Inner:", inner_df.count())
print("Left:", left_df.count())
print("Right:", right_df.count())
print("Full:", full_df.count())
```

```python
appointments_data = [
    (1, 101, 500, "CONFIRMED"),
    (2, 102, 800, "CONFIRMED"),
    (3, 101, 500, "CANCELLED"),
    (4, 103, 1200, "CONFIRMED"),
    (5, 102, 800, "PENDING"),
    (6, 101, 500, "CONFIRMED"),
    (7, 103, 1200, "CONFIRMED"),
    (8, 102, 800, "CONFIRMED"),
    (9, 104, 600, "CANCELLED"),
    (10, 104, 600, "CONFIRMED")
]

appointments_df = spark.createDataFrame(
    appointments_data,
    ["appointment_id", "doctor_id", "fees", "status"]
)

appointments_df.show()
```

```python
doctors_data = [
    (101, "Dr. Sharma", "Cardiology"),
    (102, "Dr. Khan", "Neurology"),
    (103, "Dr. Patel", "Orthopedics"),
    (104, "Dr. Verma", "Dermatology"),
    (105, "Dr. Singh", "General Medicine")
]

doctors_df = spark.createDataFrame(
    doctors_data,
    ["doctor_id", "doctor_name", "department"]
)

doctors_df.show()
```

```python
revenue_df = (
    appointments_df
    .filter(col("status") == "CONFIRMED")
    .groupBy("doctor_id")
    .agg(
        sum("fees").alias("total_revenue"),
        avg("fees").alias("average_fees"),
        count("*").alias("total_appointments")
    )
    .orderBy(desc("total_revenue"))
)

revenue_df.show()
```

```python
appointments_df.createOrReplaceTempView("appointments")
```

```python
revenue_sql_df = spark.sql("""
    select
    doctor_id,
    sum(fees) as total_revenue, 
    avg(fees) as average_fees,
    count('*') as total_appointments
    from appointments 
    where status = "CONFIRMED"
    group by doctor_id
    order by total_revenue desc
""")

revenue_sql_df.show()
```

```python
doctor_performance_df = spark.sql("""
    SELECT
        d.doctor_id,
        d.doctor_name,
        d.department,
        SUM(a.fees) AS total_revenue,
        COUNT(a.appointment_id) AS appointment_count
    FROM appointments AS a
    INNER JOIN doctors AS d
        ON a.doctor_id = d.doctor_id
    WHERE a.status = 'CONFIRMED'
    GROUP BY
        d.doctor_id,
        d.doctor_name,
        d.department
    ORDER BY total_revenue DESC
""")

doctor_performance_df.show()
```
