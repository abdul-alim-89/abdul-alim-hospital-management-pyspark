```python
import time
```

```python
def slow_double(x):
    time.sleep(0.5)      # pretend this is expensive work
    return x * 2

numbers = sc.parallelize([1, 2, 3, 4, 5, 6, 7, 8], 4)   # 8 numbers, 4 partitions

# ---------- WITHOUT cache ----------
rdd = numbers.map(slow_double)

start = time.time()
rdd.count()     # use 1
rdd.sum()       # use 2
rdd.max()       # use 3
print("Without cache:", round(time.time() - start, 1), "sec")
```

```python
def slow_double(x):
    time.sleep(0.5)      # pretend this is expensive work
    return x * 2

numbers = sc.parallelize([1, 2, 3, 4, 5, 6, 7, 8], 4)   # 8 numbers, 4 partitions

# ---------- WITHOUT cache ----------
rdd = numbers.map(slow_double)
rdd.cache()

start = time.time()
rdd.count()     # use 1
rdd.sum()       # use 2
rdd.max()       # use 3
print("Without cache:", round(time.time() - start, 1), "sec")
```

```python
lines = sc.parallelize([
    "1,Amit,25",
    "2,Neha,31",
    "3,,28",            # missing name
    "4,Raj,abc",        # age not a number
    "5,Pooja,-4",       # age out of range
    "6,Karan,40",
    "7,Sana",           # missing column
    "8,Vikas,52",
], 4)
```

```python
total_rows   = sc.accumulator(0)
invalid_rows = sc.accumulator(0)
```

```python
def validate(line):
    total_rows.add(1)
    parts = line.split(",")
    if len(parts) != 3:
        invalid_rows.add(1)
        return None
    id_, name, age = parts
    try:
        age = int(age)
    except ValueError:
        invalid_rows.add(1)
        return None
    if not name or age < 0 or age > 120:
        invalid_rows.add(1)
        return None
    return (int(id_), name, age)

valid = lines.map(validate).filter(lambda r: r is not None)

# Step 3: an action triggers the work
print("Valid rows  :", valid.collect())

# Step 4: read the accumulators on the driver, AFTER the action
print("Total rows  :", total_rows.value)     # 8
print("Invalid rows:", invalid_rows.value)   # 5

input("Open http://localhost:4040, then press Enter to exit...")
spark.stop()
```
