```python
appointments = [
    (101, 1),
    (102, 2),
    (103, 1),
    (104, 3)
]
```

```python
departments = {
    1: "Cardiology",
    2: "Neurology",
    3: "Orthopedics"
}
```

```python
rdd = sc.parallelize(appointments, 4)

result = rdd.map(
    lambda x: (
        x[0],
        departments.get(x[1], "Unknown")
    )
)

print(result.collect())
```
